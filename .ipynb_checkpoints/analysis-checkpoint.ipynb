## Introduction

This study sought to illucidate the effect of transcription factor RUNX1 on chromatin organization and gene regulation in breast cancer cells. RUNX1 is known to be an oncogene and tumor supressor however there is a gap in knowledge for how RUNX1 modulates gene regulation in a 3-dimentional setting, adding another level to its ability to ability to regulate genetic expression. To answer this question they performed ChIP-Seq to map the binding of RUNX1, RNA-seq to observe changes in gene expression, and Hi-C to observe changes in the 3-dimentional organization of the genome. Further understanding of the impact of RUNX1 on chromatin organization on the gene expression of breast cancer cells will help to further research in the genomics and cancer fields to create therapies for patients.

## Methods

### Pipeline Overview
RUNX1 ChIP-seq data were processed using Nextflow v25.04.6. Singularity containers were used for reproducibility. The pipeline processed two biological replicates of RUNX1 ChIP-seq (IP_rep1 and IP_rep2) and corresponding input controls (INPUT_rep1 and INPUT_rep2).

### Quality Control and Read Trimming
Raw sequencing reads were assessed for quality using FastQC v0.12.1. Adapter sequences and low-quality bases were trimmed using Trimmomatic v0.39 with default parameters. Quality control metrics across all samples were aggregated using MultiQC v1.15.

### Read Alignment
Following quality trimming, reads were aligned to the human reference genome (GRCh38/hg38) using Bowtie2 v2.4.5 with default parameters. A Bowtie2 index was built from the GRCh38 primary assembly genome sequence prior to alignment. SAMtools v1.21 was used to convert aligned reads to BAM format, sorted by genomic coordinate using SAMtools sort, and indexed using SAMtools index. Alignment statistics were generated using SAMtools flagstat. 

### Signal Track Generation and Visualization
Genome-wide coverage tracks were generated from sorted BAM files using deepTools bamCoverage v3.5.5, producing normalized bigWig files for visualization. Signal enrichment profiles were computed using deepTools computeMatrix and visualized using deepTools plotProfile. Sample correlation analysis was performed using deepTools multiBigwigSummary followed by deepTools plotCorrelation to assess replicate concordance.

### Peak Calling
ChIP-seq peaks were called using HOMER v4.11 findPeaks. Tag directories were created from sorted BAM files using HOMER makeTagDirectory. Peak calling was performed independently for each biological replicate using the corresponding input control as background. HOMER findPeaks was run with the following parameters: FDR threshold = 0.001, fold enrichment over input greater than or equal to 4, Poisson p-value threshold = 1.00e-04, local fold enrichment greater than or equal to 4 over a 10 kb region, and local Poisson p-value threshold = 1.00e-04.

### Identification of Reproducible Peaks
To identify high-confidence peaks present in both replicates, peak files were converted from HOMER format to BED format using HOMER pos2bed.pl. Reproducible peaks were identified using BEDTools intersect v2.31.1, requiring at least 1 bp overlap between replicates. Peaks overlapping ENCODE blacklist regions were removed using BEDTools subtract. 

### Peak Annotation
Reproducible peaks were annotated to genomic features using HOMER annotatePeaks.pl with the GENCODE v45 primary assembly annotation and the GRCh38 primary assembly genome sequence. 

### Motif Discovery
De novo motif discovery was performed on reproducible peaks using HOMER findMotifsGenome.pl to identify enriched transcription factor binding motifs within RUNX1 binding sites.

### Functional Enrichment Analysis
Unique gene names were extracted from the annotated peak file and used for functional enrichment analysis. Gene ontology and pathway enrichment were performed using Enrichr. Statistical significance was determined using Fisher's exact test with Benjamini-Hochberg FDR correction (adjusted p-value < 0.05).

### Integration with RNA-seq Data
To assess the regulatory relationship between RUNX1 binding and gene expression, annotated ChIP-seq peaks were overlapped with lists of differentially expressed genes from RNA-seq analysis. Peaks were categorized by genomic location (within +/- 5 kb of TSS or within +/- 20 kb of gene bodies) and by gene expression status (up-regulated or down-regulated). The proportion of differentially expressed genes with RUNX1 binding at each genomic location was calculated and visualized.


## Quality Control

Examining our MultiQC report we see that the general statistics section tells us a few key attributes of our data to keep in mind. First off, all of the reads have similar mapped amounts other than the control INPUT_rep2 with less than half of the reads of the other samples (10.2 million reads). It seems that something could have gone wrong in sample collection or when generating reads but since it is used for control/background it can be forgiven that it has less reads than the other samples. Another attribute we notice in the general statistics is that there are a high number of duplicates in the experimental samples IP_rep1 and IP_rep2 (89.1% and 74.3%). This is also something to keep an eye on but may be an artifact of working with ChIP-Seq data due to the fact that we are looking at specific regions that are biased to be repeated such as the transcription factor RUNX1

Next in our report we see the flagstat: read count section which tells us that what we discussed earlier, that there was a good amount of reads and alignment for each of the samples other than INPUT_rep2.

Trimmomatic shows us that we dropped a few million reads per sample, with more reads being dropped in the experimental group. This is normal and nothing to be concerned about.

Next we see the FastQC sequence counts report which shows us that there is very high duplication rates for our experimental group. This is something to keep an eye on and be concerned about but as we stated earlier, it may be due to the fact that we are looking at ChIP-seq data of a specific transcription factor. We then see the mean quality scores graph which show us a slight dip in the front and back end of the reads, which is completley to be expected in a normal experiment. These reads all maintain a score above 30 and therefore pass. The per sequence quality score graph depicts the quality of reads to catch outlier groups, however the results are nominal. The next two graphs depict GC content and show that there is a slight difference between the experimental and control groups with the control group actually failing the check. However we might expect the GC and nucleotide content to differ between the groups as with the control group we would expect there to be a close to random distribution of nucleotides/GC content while with the experimental group we are looking at specifically where RUNX1 is binding so it will definitely bias the GC content. The per base N content shows little missing base calls.

Next in the report we see a sequence duplication levels report depicting high levels of duplication for IP_rep1 and IP_rep2. Again this can be explained as normal because the duplication arrises from the fact that we are looking at RUNX1 binding sites which are bound to be more duplicated because they are a small subset of sequences. The opposite is true for the control group because they are meant to be random background. The same is shown in the overrepresented sequences sample summary, but this is expected. We additionnaly see a small number of over-represented sequences that are presumably from contamination during sample proccesing. Due to the small amount of contamination, we can safely ignore it for now. 

Overall I can safely say that the experiement was of generally high-quality and is suitable for downstream analysis. This is because any of the fails we saw during the quality check can be attributed to the structure of the experiment with the other minor details such as low read counts in the control sample being negligeble. 

## Spearman Correlation Distance between Correlation Coefficients:

![Gene Body Profile](results/correlation/correlation_spearman_heatmap.png)

I chose Spearman Correlation because ChIP-seq data is usually not normally distributed, Spearman can handle non-linear relationships caused by seq depth differences, and PCR amp bias. Spearman can also handle extreme values better from high read counts at strong binding sights. My correlation plot did not show as much of a difference between the experimental groups and control as the original paper however it still shows seperation.

## BEDTOOLS INTERSECT REPRODUCIBLE PEAKS:

I chose a reciprocal overlap approach that looks for a minimum of 50% of overlap between features in file A and file B (-f 0.50) and making the 50% overlap reciprocal, going in both directions (-r). This approach filters out false positives and artifacts while the recipricol nature prevents biases in peak differences between samples.

```python

total_up_genes = len(POS_Genes)  # Total up-regulated genes
total_down_genes = len(NEG_Genes)  # Total down-regulated genes

# For +/- 5kb of TSS
up_5kb_bound = POS_Peaks[(POS_Peaks['Distance to TSS'] < 5000) & 
                          (POS_Peaks['Distance to TSS'] > -5000)]['Gene Name'].nunique()
down_5kb_bound = NEG_Peaks[(NEG_Peaks['Distance to TSS'] < 5000) & 
                            (NEG_Peaks['Distance to TSS'] > -5000)]['Gene Name'].nunique()

# For "whole gene" - peaks that are NOT intergenic
up_gene_bound = POS_Peaks[
    ~POS_Peaks['Annotation'].str.contains('Intergenic', case=False, na=False)
]['Gene Name'].nunique()

down_gene_bound = NEG_Peaks[
    ~NEG_Peaks['Annotation'].str.contains('Intergenic', case=False, na=False)
]['Gene Name'].nunique()

# Calculate not bound using the ORIGINAL gene list totals
up_5kb_notbound = total_up_genes - up_5kb_bound
down_5kb_notbound = total_down_genes - down_5kb_bound
up_gene_notbound = total_up_genes - up_gene_bound
down_gene_notbound = total_down_genes - down_gene_bound

# Prepare data for plotting
labels = ['Up-regulated', 'Down-regulated', 'Up-regulated', 'Down-regulated']
sublabels = ['+/- 5kb of TSS', '+/- 5kb of TSS', '+/- 20kb of whole gene', '+/- 20kb of whole gene']

runx2_bound = [up_5kb_bound, down_5kb_bound, up_gene_bound, down_gene_bound]
not_bound = [up_5kb_notbound, down_5kb_notbound, up_gene_notbound, down_gene_notbound]
totals = [total_up_genes, total_down_genes, total_up_genes, total_down_genes]

# Calculate percentages
not_bound_pct = [nb/tot * 100 for nb, tot in zip(not_bound, totals)]
runx2_bound_pct = [rb/tot * 100 for rb, tot in zip(runx2_bound, totals)]

# Create the plot
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(len(labels))
width = 0.6

fig, ax = plt.subplots(figsize=(10, 6))

p1 = ax.bar(x, not_bound_pct, width, color='gray', label='Not bound')
p2 = ax.bar(x, runx2_bound_pct, width, bottom=not_bound_pct, color='#FF6B35', label='RUNX1 bound')

# Add count labels
for i, (nb, rb) in enumerate(zip(not_bound, runx2_bound)):
    ax.text(i, not_bound_pct[i]/2, str(nb), ha='center', va='center', 
            fontsize=11, fontweight='bold', color='white')
    ax.text(i, not_bound_pct[i] + runx2_bound_pct[i]/2, str(rb), ha='center', va='center', 
            fontsize=11, fontweight='bold', color='white')

ax.set_ylabel('Percentage of genes', fontsize=12)
ax.set_ylim(0, 105)
ax.set_yticks(range(0, 101, 10))
ax.set_yticklabels([f'{i}%' for i in range(0, 101, 10)])
ax.set_xticks(x)

x_labels = [f'{label}\n{sublabel}' for label, sublabel in zip(labels, sublabels)]
ax.set_xticklabels(x_labels, rotation=0, ha='center')

ax.legend(loc='upper right')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

There are some differences between my results and that of the paper. Most notable differences are found in the +/- 20kb columns with the paper showing a much larger overlap of RUNX1 bound genes than I did. The total amount of genes for each column is consistent with the paper and the amount of overlap for the two first columns is comparable, however the paper depicts a much larger overlap in the amount of RUNX1 bound genes being up/downregulated in the 20kb category while my data only depicts a negligeble increase in overlap. These differences can be explained by how the paper defines +/- 20kb of whole gene meaning if they took in a larger area around the gene they could pick up more overlap. Another way they can be explained is that the gene annotation for these peaks was done using the hg19 while we used hg38. Using a different gene annotation definitely had significant impact on the gene overlaps. Combining the analysis of both the 5kb and 20kb regions allows us to look at direct binding (5kb) and indirect effectors of the RUNX1 transcription factor. This would provide us with valuable information on off-target effects and clues into the mechanism for how RUNX1 functions. It would also allow us to differentiate between direct effects of binding by RUNX1 and that of more distant bindings.

## Sup Fig S2C Venn Diagram

```python


# from peak_stats.txt of reproducible peaks
total_rep1 = 22897
total_rep2 = 89360
shared = 5050

# Calculate unique to each replicate
unique_rep1 = total_rep1 - shared  # 17847
unique_rep2 = total_rep2 - shared  # 84310

print(f"Unique to Rep1: {unique_rep1}")
print(f"Unique to Rep2: {unique_rep2}")
print(f"Shared: {shared}")

# Create figure
fig, ax = plt.subplots(figsize=(10, 8))

# Create Venn diagram
v = venn2(subsets=(unique_rep1, unique_rep2, shared),
          set_labels=('Rep1', 'Rep2'),
          ax=ax)

# Customize colors to match the paper
v.get_patch_by_id('10').set_color('#FF6B6B')  # Rep1 only - red/coral
v.get_patch_by_id('01').set_color('#A8B5FF')  # Rep2 only - blue
v.get_patch_by_id('11').set_color('#C77DFF')  # Overlap - purple

# Adjust transparency
v.get_patch_by_id('10').set_alpha(0.7)
v.get_patch_by_id('01').set_alpha(0.6)
v.get_patch_by_id('11').set_alpha(0.7)

# Customize text - labels
for text in v.set_labels:
    text.set_fontsize(28)
    text.set_fontweight('bold')
    
# Customize text - numbers
for text in v.subset_labels:
    text.set_fontsize(22)
    text.set_fontweight('bold')

# Remove axis
ax.axis('off')

plt.show()
```

There are signficant differences between my called peaks and that of the paper. For instance I have 17,847 and 84,310 called peaks for Rep1 and 2 respectively. Meanwhile the paper has only a fraction of this amount with 3,983 and 10,465 peaks for Rep1 and 2 respectively. The shared peaks is more similar with the original paper having 3,466 peaks to my 5,050. The discrepancies between my called peaks and that of the original paper likely has to do with the differences in how the data was filtered, as in the paper had more aggressive filtering when compared to my approach. Another difference could stem from how the paper called peaks. I called peaks with a regressive 50% overlap but the paper may have been more stringent.



## Analyze Annotated Peaks

```python

# Load annotated peaks file
peaks = pd.read_csv('results/peak_annotation/annotated_peaks.txt', sep='\t')

print(f"Total peaks: {len(peaks)}")

# Extract unique gene names
genes = peaks['Gene Name'].dropna().unique().tolist()

print(f"Total unique genes with RUNX1 binding: {len(genes)}")
print(f"First 10 genes: {genes[:10]}")

# Save gene list
with open('RUNX1_target_genes.txt', 'w') as f:
    f.write('\n'.join(genes))
```

```python

# Filter for promoter peaks
promoter_peaks = peaks[peaks['Annotation'].str.contains('promoter', case=False, na=False)]

# Extract unique promoter-bound genes
promoter_genes = promoter_peaks['Gene Name'].dropna().unique().tolist()

print(f"Total promoter-bound genes: {len(promoter_genes)}")
print(f"First 10 genes: {promoter_genes[:10]}")

# Save to text file
with open('RUNX1_promoter_bound_genes.txt', 'w') as f:
    f.write('\n'.join(promoter_genes))
```

From the annotated peaks data I filtered for promoter-bound genes. These genes would show us direct transcriptional targets of RUNX1, removing any indirect regulatory relathionships, and allowing us to focus on a few key pathways. We see in the reactome pathways 2024, a strong enrichment for histone modifications as well in chromatin modifying enzymes, and chromatin organization. This is notable because RUNX1 is a master transcription factor that utilizees chromatin remodeling complexes to regulate gene expression during cell development. In BioPlanet 2019 we see pathways for cell cycle and G2/M checkpoints which shows its direct role in cell cycle progression and its involvment in breast cancer development as an oncogene/tumor supressor. This is further supported by Acute myeloid leukemia as seen in the KEGG 2021 Human pathway list. In this pathway we also see RUNX1's involvment in metabolic pathways; glycolysis, citrate cycle, and lysine degradation. Its involvment in coordinating cellular stress response (mTOR signaling) further illucidates its far reaching impact on gene expression and its utility in cancer treatment and research.
